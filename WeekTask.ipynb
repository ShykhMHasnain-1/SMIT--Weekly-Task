<h1>Calculate Frequency And Percentage Distributions For Diffrent Columns</h1>

```python
for col in Data.columns:
    category_count = Data[col].value_counts()
    per_cat_count = ((category_count/len(Data[col]))*100)
    result = pd.DataFrame({
        "Frequency": category_count,
        "Percentage": per_cat_count.round(2)
    })
    print(result)
```

<h1>Calculate The Number Of Students In Each Adaptability Category</h1>

```python
adpt_level_count = Data["Adaptivity Level"].value_counts()

result = pd.DataFrame({
    "Adaptivity Levels": adpt_level_count.index,
    "Frequency": adpt_level_count.values,
    "Percentage": (adpt_level_count / len(Data["Adaptivity Level"]) * 100).round(2)
})

print(result)
```

<h1>Most Common Adaptability Levels</h1>

```python
adpt_level_count.idxmax()
```

<h1>Least Common Adaptability Levels</h1>

```python
adpt_level_count.idxmin()
```

<h1>Financial Condition Analysis</h1>

```python
finance_level_count = Data["Financial Condition"].value_counts()

result = pd.DataFrame({
    "Adaptivity Levels": finance_level_count.index,
    "Frequency": finance_level_count.values,
    "Percentage": (finance_level_count / len(Data["Financial Condition"]) * 100).round(2)
})

print(result)
```
